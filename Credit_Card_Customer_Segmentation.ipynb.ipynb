```python

import pandas as pd

df = pd.read_csv('/content/CC GENERAL.csv')

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)

display(df.head())
```

```python

print("Dataset information:")
df.info()
```

```python

# Check missing values in each column
print("Missing values before cleaning:")
print(df.isnull().sum())

# Check duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())

# Fill missing values in numeric columns with their median
df['MINIMUM_PAYMENTS'] = df['MINIMUM_PAYMENTS'].fillna(
    df['MINIMUM_PAYMENTS'].median()
)

df['CREDIT_LIMIT'] = df['CREDIT_LIMIT'].fillna(
    df['CREDIT_LIMIT'].median()
)

# Check missing values after cleaning
print("\nMissing values after cleaning:")
print(df.isnull().sum())
```

```python

# Display summary statistics for numeric columns
df.describe().round(2)
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

# Select important customer behavior features
features = [
    'BALANCE',
    'PURCHASES',
    'CASH_ADVANCE',
    'CREDIT_LIMIT',
    'PAYMENTS'
]

# Plot distributions
plt.figure(figsize=(15, 8))

for i, col in enumerate(features, 1):
    plt.subplot(2, 3, i)
    sns.histplot(df[col], bins=30, kde=True)
    plt.title(col)
    plt.xlabel(col)
    plt.ylabel('Customer Count')

plt.tight_layout()
plt.show()
```

```python

# Remove the customer ID because it is only an identifier
X = df.drop(columns=['CUST_ID'])

# Keep numeric columns for clustering
X = X.select_dtypes(include=['number'])

print("Features selected:", X.shape[1])
print("Dataset shape:", X.shape)

display(X.head())
```

```python

from sklearn.preprocessing import StandardScaler

# Standardize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Original feature shape:", X.shape)
print("Scaled feature shape:", X_scaled.shape)

print("\nFirst 5 scaled rows:")
print(X_scaled[:5])
```

```python

from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

wcss = []

for k in range(1, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

# Plot the Elbow curve
plt.figure(figsize=(8, 5))
plt.plot(range(1, 11), wcss, marker='o')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('WCSS')
plt.title('Elbow Method for Optimal K')
plt.xticks(range(1, 11))
plt.grid(True)
plt.show()
```

```python

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)

    print(f"K = {k}, Silhouette Score = {score:.4f}")
```

```python

from sklearn.cluster import KMeans

# Choose the number of clusters
k = 4

# Create and train the model
kmeans = KMeans(
    n_clusters=k,
    random_state=42,
    n_init=10
)

labels = kmeans.fit_predict(X_scaled)

# Add cluster labels to the original dataset
df_clustered = df.copy()
df_clustered['Cluster'] = labels

print("Model trained successfully!")
print("\nCustomers in each cluster:")
print(df_clustered['Cluster'].value_counts().sort_index())
```

```python

cluster_summary = df_clustered.groupby('Cluster')[
    ['BALANCE', 'PURCHASES', 'CASH_ADVANCE',
     'CREDIT_LIMIT', 'PAYMENTS']
].mean().round(2)

display(cluster_summary)
```

```python

from sklearn.decomposition import PCA
import pandas as pd
import matplotlib.pyplot as plt

# Reduce scaled features to two dimensions
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)

# Create a visualization dataset
plot_df = pd.DataFrame({
    'PCA1': X_pca[:, 0],
    'PCA2': X_pca[:, 1],
    'Cluster': df_clustered['Cluster'].values
})

print("PCA completed!")
print("Variance explained:", round(pca.explained_variance_ratio_.sum() * 100, 2), "%")
```

```python

plt.figure(figsize=(10, 6))

for cluster in sorted(plot_df['Cluster'].unique()):
    subset = plot_df[plot_df['Cluster'] == cluster]
    plt.scatter(
        subset['PCA1'],
        subset['PCA2'],
        label=f'Cluster {cluster}',
        alpha=0.6,
        s=20
    )

plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('Credit Card Customer Segmentation (K-Means)')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python

from sklearn.decomposition import PCA
import pandas as pd
import matplotlib.pyplot as plt

# Reduce scaled data to 3 principal components
pca = PCA(n_components=3)
X_pca_3d = pca.fit_transform(X_scaled)

plot_df_3d = pd.DataFrame({
    'PCA1': X_pca_3d[:, 0],
    'PCA2': X_pca_3d[:, 1],
    'PCA3': X_pca_3d[:, 2],
    'Cluster': df_clustered['Cluster'].values
})

print(
    "Variance explained by 3 components:",
    round(pca.explained_variance_ratio_.sum() * 100, 2),
    "%"
)
```

```python

from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(11, 8))
ax = fig.add_subplot(111, projection='3d')

scatter = ax.scatter(
    plot_df_3d['PCA1'],
    plot_df_3d['PCA2'],
    plot_df_3d['PCA3'],
    c=plot_df_3d['Cluster'],
    cmap='tab10',
    s=12,
    alpha=0.7
)

ax.set_xlabel('PCA Component 1')
ax.set_ylabel('PCA Component 2')
ax.set_zlabel('PCA Component 3')
ax.set_title('3D Credit Card Customer Segmentation')

legend = ax.legend(
    *scatter.legend_elements(),
    title='Cluster'
)
ax.add_artist(legend)

plt.show()
```

```python

from sklearn.metrics import silhouette_score, davies_bouldin_score

# Evaluate the trained K-Means model
labels = df_clustered['Cluster']

sil_score = silhouette_score(X_scaled, labels)
db_score = davies_bouldin_score(X_scaled, labels)

print("Model Evaluation Results")
print("------------------------")
print("Number of clusters:", df_clustered['Cluster'].nunique())
print("Silhouette Score:", round(sil_score, 4))
print("Davies-Bouldin Score:", round(db_score, 4))
```

```python

# Calculate average customer behavior by cluster
features = [
    'BALANCE',
    'PURCHASES',
    'CASH_ADVANCE',
    'CREDIT_LIMIT',
    'PAYMENTS',
    'PURCHASES_FREQUENCY'
]

cluster_summary = df_clustered.groupby('Cluster')[features].mean().round(2)

print("CUSTOMER SEGMENTATION SUMMARY")
display(cluster_summary)

print("\nCUSTOMER COUNT BY CLUSTER")
print(df_clustered['Cluster'].value_counts().sort_index())
```

```python

# Compare each cluster's averages with the overall customer averages
overall_avg = df_clustered[features].mean()

for cluster in sorted(df_clustered['Cluster'].unique()):
    cluster_avg = cluster_summary.loc[cluster]

    print(f"\nCluster {cluster}")
    print("-" * 30)

    for feature in features:
        if cluster_avg[feature] > overall_avg[feature]:
            print(f"{feature}: Above overall average")
        elif cluster_avg[feature] < overall_avg[feature]:
            print(f"{feature}: Below overall average")
        else:
            print(f"{feature}: Near overall average")
```
